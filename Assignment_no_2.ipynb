```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
```

```python
iris = load_iris()
x = iris.data
y = iris.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_scaled, y, test_size=0.3, random_state=42, stratify=y)
```

```python
lr_no_pca = LogisticRegression(max_iter=500)
lr_no_pca.fit(x_train, y_train)
y_pred_lr_no_pca = lr_no_pca.predict(x_test)
acc_lr_no_pca = accuracy_score(y_test, y_pred_lr_no_pca)
print("Logistic Regression Accuracy (without PCA):", acc_lr_no_pca)
```

```python
knn_no_pca = KNeighborsClassifier(n_neighbors=5)
knn_no_pca.fit(x_train, y_train)
y_pred_knn_no_pca = knn_no_pca.predict(x_test)
acc_knn_no_pca = accuracy_score(y_test, y_pred_knn_no_pca)
print("KNN Accuracy (without PCA):", acc_knn_no_pca)
```

```python
pca = PCA(n_components=2)
x_pca = pca.fit_transform(x_scaled)
```

```python
x_train_pca, x_test_pca, y_train_pca, y_test_pca = train_test_split(x_pca, y, test_size=0.3, random_state=42, stratify=y)
```

```python
lr_pca = LogisticRegression(max_iter=500)
lr_pca.fit(x_train_pca, y_train_pca)
y_pred_lr_pca = lr_pca.predict(x_test_pca)
acc_lr_pca = accuracy_score(y_test_pca, y_pred_lr_pca)
print("Logistic Regression Accuracy (with PCA):", acc_lr_pca)
```

```python
knn_pca = KNeighborsClassifier(n_neighbors=5)
knn_pca.fit(x_train_pca, y_train_pca)
y_pred_knn_pca = knn_pca.predict(x_test_pca)
acc_knn_pca = accuracy_score(y_test_pca, y_pred_knn_pca)
print("KNN Accuracy (with PCA):", acc_knn_pca)
```

```python
matrix = confusion_matrix(y_test_pca, y_pred_knn_pca)
print("Confusion Matrix (KNN with PCA):\n", matrix)

display = ConfusionMatrixDisplay(confusion_matrix=matrix, display_labels=iris.target_names)
display.plot()
plt.title("KNN Confusion Matrix with PCA")
plt.show()
```

```python
print("\n--- Accuracy Comparison ---")
print(f"Logistic Regression: Without PCA = {acc_lr_no_pca:.3f}, With PCA = {acc_lr_pca:.3f}")
print(f"KNN:                 Without PCA = {acc_knn_no_pca:.3f}, With PCA = {acc_knn_pca:.3f}")
```
