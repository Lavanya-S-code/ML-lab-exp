 CREATING AN ARRAY BY USING LIST


```python
import numpy as np
arr=np.array([1,2,3,4,5])
print(arr)
print(type(arr))
```

CREATING AN ARRAY BY USING TUPLE

```python
import numpy as np
arr=np.array((1,2,3,4,5))
print(arr)
```

TO CHECK THE DIMENSIONAL OF THE ARRAY


```python
import numpy as np
arr=np.array([1,2,3,4,5])
print(arr)
print("Dimension of the array:",arr.ndim)
```
