```python
from hmmlearn import hmm
import numpy as np
```

```python
states = ["Sunny", "Rainy"]
observations = ["Walk", "Shop", "Clean"]
```

```python
start_prob = np.array([0.7, 0.3])
trans_mat = np.array([
    [0.8, 0.2],   # Sunny -> Sunny, Rainy
    [0.3, 0.7]    # Rainy -> Sunny, Rainy
])
emit_mat = np.array([
    [0.6, 0.3, 0.1],  # Sunny emits Walk, Shop, Clean
    [0.1, 0.4, 0.5]   # Rainy emits Walk, Shop, Clean
])
```

```python
model = hmm.MultinomialHMM(n_components=2, n_iter=100)
model.startprob_ = start_prob
model.transmat_ = trans_mat
model.emissionprob_ = emit_mat
model.n_trials = 1
```

```python
# Observation Sequence (correct 2D format)
obs_seq = ([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0]
])
```

```python
# Decode using Viterbi Algorithm
logprob_viterbi, hidden_states_viterbi = model.decode(obs_seq, algorithm="viterbi")
```

```python
print("=== VITERBI ALGORITHM ===")
print("Observed sequence: ", [observations[i[0]] for i in obs_seq])
print("Most likely hidden states: ", [states[i] for i in hidden_states_viterbi])
print("Log probability (Viterbi):", logprob_viterbi)

```

```python
logprob_forward = model.score(obs_seq)  # log-likelihood of the observation sequence
print("\n=== FORWARD ALGORITHM ===")
print("Log probability (Forward):", logprob_forward)
```

```python
print("\n=== COMPARISON ===")
print(f"Viterbi Log Probability: {logprob_viterbi:.4f}")
print(f"Forward Log Probability: {logprob_forward:.4f}")
```
