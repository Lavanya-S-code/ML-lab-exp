```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
iris = load_iris()
x = iris.data
y = iris.target
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
y_kmeans = kmeans.fit_predict(x_scaled)

print(" K-MEANS CLUSTERING RESULTS \n")
print("K-Means Confusion Matrix:")
cm_kmeans = confusion_matrix(y, y_kmeans)
print(cm_kmeans)
```

```python
display_kmeans = ConfusionMatrixDisplay(confusion_matrix=cm_kmeans, display_labels=iris.target_names)
display_kmeans.plot()
plt.title("K-Means vs True Labels")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(x_scaled[:, 0], x_scaled[:, 1], c=y_kmeans, cmap='rainbow')
plt.title("K-Means Clustering on Iris Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
hier = AgglomerativeClustering(n_clusters=3)
y_hier = hier.fit_predict(x_scaled)

print("\HIERARCHICAL CLUSTERING RESULTS\n")
print("Hierarchical Confusion Matrix:")
cm_hier = confusion_matrix(y, y_hier)
print(cm_hier)
```

```python
display_hier = ConfusionMatrixDisplay(confusion_matrix=cm_hier, display_labels=iris.target_names)
display_hier.plot()
plt.title("Hierarchical Clustering vs True Labels")
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.scatter(x_scaled[:, 0], x_scaled[:, 1], c=y_hier, cmap='rainbow')
plt.title("Hierarchical Clustering on Iris Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
link = linkage(x_scaled, method='ward')
dendrogram(link)
plt.title("Dendrogram for Hierarchical Clustering")
plt.xlabel("Samples")
plt.ylabel("Distance")
plt.show()
```
